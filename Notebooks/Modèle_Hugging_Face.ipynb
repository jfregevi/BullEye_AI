```python
from google.colab import drive
import pandas as pd
import random
import uuid
from datetime import datetime, timedelta

# 1️⃣ Monter Google Drive
drive.mount('/content/drive')

# Cryptomonnaies possibles avec plages de montants reçus
cryptos = {
    "BTC": (0.0005, 0.005),
    "ETH": (0.01, 0.2),
    "LTC": (0.1, 2),
    "XRP": (5, 200)
}

def generate_random_transactions(num_rows, output_file):
    rows = []

    for _ in range(num_rows):
        # Choisir une crypto aléatoire
        crypto = random.choice(list(cryptos.keys()))
        min_val, max_val = cryptos[crypto]

        received_amount = round(random.uniform(min_val, max_val), 8)
        sent_amount = round(random.uniform(20, 500), 2)
        fee_amount = round(sent_amount * random.uniform(0.005, 0.02), 2)
        description = random.choice(["Buy", "Sell"])

        # Générer une date aléatoire
        date = datetime(2025, 4, 19, 13, 0, 0) + timedelta(seconds=random.randint(0, 3600))
        date_str = date.isoformat() + "Z"

        rows.append({
            "type": "Trade",
            "date": date_str,
            "timezone": "GMT",
            "received_amount": received_amount,
            "received_currency": crypto,
            "sent_amount": sent_amount,
            "sent_currency": "EUR",
            "fee_amount": fee_amount,
            "fee_currency": "EUR",
            "description": description,
            "address": "",
            "transaction_hash": "",
            "external_id": str(uuid.uuid4())
        })
```

```python
def calcul_positions(df):
    positions = {}
    dépenses = 0

    for _, row in df.iterrows():
        desc = str(row['description']).lower()

        # --- Cas BUY ---
        if "buy" in desc:
            asset = row['received_currency']
            qty = row['received_amount']
            cost = row['sent_amount']  # EUR payé
            dépenses += row['sent_amount'] + row['fee_amount']
            if asset not in positions:
                positions[asset] = {"qty": 0, "cost": 0}
            positions[asset]["qty"] += qty
            positions[asset]["cost"] += cost

        # --- Cas SELL ---
        elif "sell" in desc:
            asset = row['sent_currency']
            qty = row['sent_amount']
            if asset not in positions:
                positions[asset] = {"qty": 0, "cost": 0}
            avg_cost = positions[asset]["cost"] / positions[asset]["qty"] if positions[asset]["qty"] > 0 else 0
            positions[asset]["qty"] -= qty
            positions[asset]["cost"] -= avg_cost * qty

        # --- Cas SWAP ---
        elif "swap" in desc:
            asset_sent = row['sent_currency']
            qty_sent = row['sent_amount']
            asset_recv = row['received_currency']
            qty_recv = row['received_amount']

            # Partie vente
            if asset_sent not in positions:
                positions[asset_sent] = {"qty": 0, "cost": 0}
            avg_cost = positions[asset_sent]["cost"] / positions[asset_sent]["qty"] if positions[asset_sent]["qty"] > 0 else 0
            positions[asset_sent]["qty"] -= qty_sent
            positions[asset_sent]["cost"] -= avg_cost * qty_sent

            # Partie achat
            if asset_recv not in positions:
                positions[asset_recv] = {"qty": 0, "cost": 0}
            positions[asset_recv]["qty"] += qty_recv
            positions[asset_recv]["cost"] += avg_cost * qty_sent

    # --- Conversion en DataFrame ---
    result = []
    for asset, data in positions.items():
        if data["qty"] <= 0:
            continue
        avg_cost = data["cost"] / data["qty"] if data["qty"] > 0 else 0
        try:
            prix_actuel = get_price_binance(asset + "EUR")
        except:
            prix_actuel = None
        plus_value = (prix_actuel - avg_cost) * data["qty"] if prix_actuel else None
        result.append({
            "Asset": asset,
            "Qty": data["qty"],
            "AvgCost": avg_cost,
            "PrixActuel": prix_actuel,
            "PlusValue": plus_value
        })

    return pd.DataFrame(result), pd.DataFrame(result)["PlusValue"].sum(), dépenses

# Lecture du CSV
df_pos, total_plus_value, dépenses = calcul_positions(df1)

print(df_pos)
print("\nTotal plus-value :", total_plus_value)
print("Voici le montant investi : ", dépenses)
Pourcentage = total_plus_value/dépenses*100
print("Cela correspond donc à une évolution de",Pourcentage, "%")
```

```python
import requests

def get_price_binance(symbol):
    """
    symbol : par ex. 'BTCEUR', 'ETHUSD', 'SOLEUR'
    """
    url = f"https://api.binance.com/api/v3/ticker/price?symbol={symbol}"
    r = requests.get(url)
    data = r.json()
    return float(data['price'])

# Exemple d'utilisation
btc_price = get_price_binance("BTCEUR")
eth_price = get_price_binance("ETHEUR")
sol_price = get_price_binance("SOLEUR")

print("BTC-EUR :", btc_price)
print("ETH-EUR :", eth_price)
print("SOL-EUR :", sol_price)
```

```python
from datetime import datetime

# --- Classe de base pour un agent ---
class Agent:
    def __init__(self, name):
        self.name = name
        self.memory = {}

    def think1(self, prompt):
        inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
        outputs = model.generate(
            **inputs,
            max_new_tokens=20,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id  # sécurité
        )
        # Supprimer la partie du prompt dans la sortie
        generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]
        response = tokenizer.decode(generated_tokens, skip_special_tokens=True).strip()
        return response

    def think2(self, prompt):
        inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id  # sécurité
        )
        # Supprimer la partie du prompt dans la sortie
        generated_tokens = outputs[0][inputs["input_ids"].shape[1]:]
        response = tokenizer.decode(generated_tokens, skip_special_tokens=True).strip()
        return response


# --- Agent questionnaire ---
class QuestionnaireAgent(Agent):
    def ask_user(self):
        print("Pour mieux répondre à vos attentes, merci de répondre aux questions suivantes :")

        questions = [
            "Quel est votre objectif financier ? (ex: croissance, sécurité, revenu) : ",
            "Quel est votre niveau de tolérance au risque ? (faible/moyen/élevé) : ",
            "Quelle est la durée prévue de l'investissement ? (court/moyen/long terme) : "
        ]
        for q in questions:
            answer = input(f"{q} ")
            self.memory[q] = answer
        return self.memory

def get_market_data():
      assets = ["BTCEUR", "ETHEUR", "SOLEUR"]
      data = {}
      for asset in assets:
          price = get_price_binance(asset)
          data[asset] = price
      return data

# --- Agent conseil ---
class AdvisorAgent(Agent):
    def recommend(self, user_info):
        market_data = get_market_data()

        # Format des données pour le modèle
        today = datetime.now().strftime("%Y-%m-%d %H:%M")
        context = (
            f"Nous sommes le {today}.\n"
            "Voici les prix actuels des cryptos (en EUR) :\n"
        )
        for asset, price in market_data.items():
            context += f"- {asset} : {price:.2f} €\n"

        context += (
            "Voici les prix actuels de mes positions (en EUR) :\n"
        )

        context += (df1.to_string(index=False))

        context += "\nFacteurs utilisateur :\n"
        for k, v in user_info.items():
            context += f"- {k} : {v}\n"

        context += (
            "\nAnalyse ces données de marché et donne-moi un conseil "
            "simple d'investissement à court terme pour un particulier, "
            "en précisant si tu recommandes d'acheter, de vendre ou de conserver.\n"
            "Réponds en français, de manière concise."
        )
        return self.think2(context)


# --- Agent pour calculer les plus-values ---
class PortfolioAgent(Agent):
    def calculate_profits(self, df):
        df_pos, total_plus_value, dépenses = calcul_positions(df)
        print(df_pos)
        print("\nTotal plus-value :", total_plus_value)
        print("Voici le montant investi : ", dépenses)
        Pourcentage = total_plus_value/dépenses*100
        print("Cela correspond donc à une évolution de",Pourcentage, "%")


class OrchestratorAgent:
    def __init__(self, q_agent, a_agent, p_agent):
        self.q_agent = q_agent
        self.a_agent = a_agent
        self.p_agent = p_agent

    def classify_request(self, user_input):
        prompt = (
            "Ta tâche est de classifier la demande de l'utilisateur selon ces règles EXCLUSIVES :\n"
            "1. Si l'utilisateur demande à calculer ou connaître ses gains, pertes, bénéfices, rendement ou plus-value sur son portefeuille, réponds strictement par 'plus-value'.\n"
            "2. Si l'utilisateur demande un avis, une stratégie, une recommandation ou un conseil en investissement, réponds strictement par 'conseil'.\n"
            "3. Dans tous les autres cas, réponds strictement par 'Je n'ai pas les compétences requises.'.\n\n"
            "⚠️ Tu dois répondre uniquement par UNE des trois chaînes de texte ci-dessus, sans rien ajouter.\n\n"
            f"Maintenant, réponds pour la demande : '{user_input}'\n"
        )

        response = self.q_agent.think1(prompt).lower()
        print("DEBUG RESPONSE:", repr(response))  # 👈 ajout debug
        response = response.lower().strip()


        if "plus-value" in response:
            return "plus-value"
        elif "conseil" in response:
            return "conseil"
        else:
            return "inconnu"

    def chat(self, market_data, portfolio):
        print("Bonjour ! Je suis votre assistant crypto. Tapez 'quitter' pour sortir.")
        while True:
            user_input = input("\nVous : ")
            if user_input.lower() == "quitter":
                print("Au revoir 👋")
                break

            task = self.classify_request(user_input)
            if task == "inconnu":
                print("Désolé, je ne sais pas répondre à cette demande.")
                continue

            if task == "conseil":
                user_info = self.q_agent.ask_user()
                advice = self.a_agent.recommend(user_info)
                print("\n=== CONSEIL D'INVESTISSEMENT ===")
                print(advice)

            elif task == "plus-value":
                file_path = input("Donnez le chemin de votre fichier CSV de portfolio : ")
                df = pd.read_excel(file_path)
                print("\n=== VOS PLUS-VALUES ===")
                return self.p_agent.calculate_profits(df)


q_agent = QuestionnaireAgent("QuestionnaireAgent")
a_agent = AdvisorAgent("AdvisorAgent")
p_agent = PortfolioAgent("PortfolioAgent")

orchestrator = OrchestratorAgent(q_agent, a_agent, p_agent)
orchestrator.chat(market_data, portfolio)
```
