```python
import numpy as np
import pandas as pd
import datetime
import matplotlib.pyplot as plt
import yfinance as yf
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (Input, Dense, Dropout, Flatten, Conv1D,
                                     MaxPooling1D, GlobalAveragePooling1D, LSTM, GRU,
                                     BatchNormalization, Softmax, Lambda, Bidirectional, Concatenate, Attention)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import time
from tcn import TCN
```

```python
# ===============================
# 1. Récupération du ticker
# ===============================
ticker = input("Entrez le ticker de l'action ou crypto (ex: AAPL, TSLA, BTC-USD): ")

end_date = datetime.date.today()
start_date = end_date - datetime.timedelta(days=10*365)

data = yf.download(ticker, start=start_date, end=end_date, progress=False).dropna()
data.sort_index(inplace=True)
df = data
print(df.head())

# ===============================
# 2. Création des séquences
# ===============================
def create_windows(df, window_size, forecast_horizon):
    X, y = [], []
    for i in range(len(df) - window_size - forecast_horizon + 1):
        window = df.iloc[i:i+window_size][["Open","High","Low","Close","Volume"]].values
        target = df.iloc[i + window_size + forecast_horizon - 1]["Close"]
        X.append(window)
        y.append(target)
    return np.array(X), np.array(y)

window_size = 20
forecast_horizon = 1
num_features = 5

X, y = create_windows(df, window_size, forecast_horizon)

# ===============================
# 3. Normalisation
# ===============================
X_scaled = np.zeros_like(X)
scalers_X = {}
for i in range(num_features):
    scalers_X[i] = MinMaxScaler()
    X_scaled[:,:,i] = scalers_X[i].fit_transform(X[:,:,i])

scaler_y = MinMaxScaler()
y_scaled = scaler_y.fit_transform(y.reshape(-1,1))

# ===============================
# 4. Split train/test
# ===============================
split = int(0.95 * len(X_scaled))
X_train, X_test = X_scaled[:split], X_scaled[split:]
y_train, y_test = y_scaled[:split], y_scaled[split:]

# ===============================
# 5. Modèles à tester (20 modèles)
# ===============================
models = {}

# 1. Dense simple
models['Dense-1'] = Sequential([
    Flatten(input_shape=(window_size, num_features)),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(1)
])

# 2. Dense plus profond
models['Dense-2'] = Sequential([
    Flatten(input_shape=(window_size, num_features)),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(1)
])

# 3. CNN simple
models['CNN-1'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size,num_features)),
    GlobalAveragePooling1D(),
    Dense(32, activation='relu'),
    Dense(1)
])

# 4. CNN profond
models['CNN-2'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size,num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    GlobalAveragePooling1D(),
    Dense(32, activation='relu'),
    Dense(1)
])

# 5. LSTM simple
models['LSTM-1'] = Sequential([
    LSTM(64, input_shape=(window_size, num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 6. LSTM profond
models['LSTM-2'] = Sequential([
    LSTM(128, return_sequences=True, input_shape=(window_size, num_features)),
    Dropout(0.2),
    LSTM(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# 7. GRU simple
models['GRU-1'] = Sequential([
    GRU(64, input_shape=(window_size,num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 8. GRU profond
models['GRU-2'] = Sequential([
    GRU(128, return_sequences=True, input_shape=(window_size,num_features)),
    Dropout(0.2),
    GRU(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# 9. BiLSTM
models['BiLSTM'] = Sequential([
    Bidirectional(LSTM(64, return_sequences=False), input_shape=(window_size,num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 10. BiGRU
models['BiGRU'] = Sequential([
    Bidirectional(GRU(64, return_sequences=False), input_shape=(window_size,num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 11. CNN + LSTM
models['CNN-LSTM'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    MaxPooling1D(2),
    LSTM(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# 12. CNN + GRU
models['CNN-GRU'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    MaxPooling1D(2),
    GRU(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# 13. CNN + BiLSTM
models['CNN-BiLSTM'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    MaxPooling1D(2),
    Bidirectional(LSTM(64, return_sequences=False)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 14. CNN + BiGRU
models['CNN-BiGRU'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    MaxPooling1D(2),
    Bidirectional(GRU(64, return_sequences=False)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 15. CNN profond + BiLSTM
models['CNN2-BiLSTM'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    Conv1D(64,3,padding='same',activation='relu'),
    MaxPooling1D(2),
    Bidirectional(LSTM(64)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 16. CNN profond + BiGRU
models['CNN2-BiGRU'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    Conv1D(64,3,padding='same',activation='relu'),
    MaxPooling1D(2),
    Bidirectional(GRU(64)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 17. CNN + LSTM + Dropout fort
models['CNN-LSTM-D'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    LSTM(64),
    Dropout(0.4),
    Dense(32, activation='relu'),
    Dense(1)
])

# 18. LSTM + Dense profond
models['LSTM-Dense'] = Sequential([
    LSTM(64, input_shape=(window_size,num_features)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

# 19. GRU + Dense profond
models['GRU-Dense'] = Sequential([
    GRU(64, input_shape=(window_size,num_features)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(1)
])

# 20. LSTM + GRU empilés
models['LSTM-GRU'] = Sequential([
    LSTM(64, return_sequences=True, input_shape=(window_size,num_features)),
    GRU(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# ===============================
# 6. Entraînement et benchmark + RMSE
# ===============================
results = {}

for name, model in models.items():
    print(f"=== Entraînement du modèle {name} ===")
    
    model.compile(optimizer=Adam(0.001), loss='mse', metrics=['mae'])
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-5)
    
    history = model.fit(
        X_train, y_train,
        epochs=100,
        batch_size=32,
        validation_data=(X_test, y_test),
        callbacks=[early_stop, reduce_lr],
        verbose=0
    )
    
    y_pred_scaled = model.predict(X_test)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_test_real = scaler_y.inverse_transform(y_test)
    
    y_pred_aligned = np.full_like(y_test_real, np.nan)
    if forecast_horizon <= len(y_pred):
        y_pred_aligned[forecast_horizon-1:] = y_pred[:len(y_pred)-forecast_horizon+1]
    
    rmse = np.sqrt(mean_squared_error(y_test_real[forecast_horizon-1:], y_pred_aligned[forecast_horizon-1:]))
    results[name] = rmse
    
    # Plot prédictions (optionnel si beaucoup de modèles)
    plt.figure(figsize=(12,4))
    plt.plot(y_test_real, label='Prix réel')
    plt.plot(y_pred_aligned, label=f'Prix prédit ({name})')
    plt.title(f'{ticker} - {name} - RMSE={rmse:.2f}')
    plt.xlabel('Index')
    plt.ylabel('Prix')
    plt.legend()
    plt.grid(True)
    plt.show()

# ===============================
# 7. Graphique comparatif RMSE
# ===============================
plt.figure(figsize=(12,5))
names = list(results.keys())
values = [results[n] for n in names]
plt.barh(names, values, color='skyblue')
plt.xlabel("RMSE")
plt.title(f"Comparaison des RMSE pour {ticker}")
plt.gca().invert_yaxis()  # le meilleur en haut
plt.grid(axis='x')
plt.show()

# ===============================
# 8. Meilleur modèle + prochaine prédiction
# ===============================
best_model_name = min(results, key=results.get)
print(f"Meilleur modèle : {best_model_name} avec RMSE = {results[best_model_name]:.4f}")

best_model = models[best_model_name]
last_window = X_scaled[-1].reshape(1, window_size, num_features)
pred_scaled = best_model.predict(last_window)
pred_close = scaler_y.inverse_transform(pred_scaled.reshape(-1,1))[0][0]

print(f"Prix prédit pour {ticker} dans {forecast_horizon} jours : {pred_close:.2f}")
```

# Comparaison des meilleurs modèles du point précédent

```python
# ===============================
# 1. Récupération du ticker
# ===============================
ticker = input("Entrez le ticker de l'action ou crypto (ex: AAPL, TSLA, BTC-USD): ")

end_date = datetime.date.today()
start_date = end_date - datetime.timedelta(days=10*365)

data = yf.download(ticker, start=start_date, end=end_date, progress=False).dropna()
data.sort_index(inplace=True)
df = data
print(df.head())

# ===============================
# 2. Création des séquences
# ===============================
def create_windows(df, window_size, forecast_horizon):
    X, y = [], []
    for i in range(len(df) - window_size - forecast_horizon + 1):
        window = df.iloc[i:i+window_size][["Open","High","Low","Close","Volume"]].values
        target = df.iloc[i + window_size + forecast_horizon - 1]["Close"]
        X.append(window)
        y.append(target)
    return np.array(X), np.array(y)

window_size = 20
forecast_horizon = 1
num_features = 5

X, y = create_windows(df, window_size, forecast_horizon)

# ===============================
# 3. Normalisation
# ===============================
X_scaled = np.zeros_like(X)
scalers_X = {}
for i in range(num_features):
    scalers_X[i] = MinMaxScaler()
    X_scaled[:,:,i] = scalers_X[i].fit_transform(X[:,:,i])

scaler_y = MinMaxScaler()
y_scaled = scaler_y.fit_transform(y.reshape(-1,1))

# ===============================
# 4. Split train/test
# ===============================
split = int(0.95 * len(X_scaled))
X_train, X_test = X_scaled[:split], X_scaled[split:]
y_train, y_test = y_scaled[:split], y_scaled[split:]

# ===============================
# 5. Modèles à tester (20 modèles)
# ===============================
models = {}

# ===============================================================================
# 🚀 20 MODÈLES AMÉLIORÉS (Basés sur GRU, BiGRU, CNN2-BiRNN, GRU-Dense)
# ===============================================================================

# --- Améliorations GRU (Meilleur modèle initial) et GRU-Dense ---

# 1. GRU-1 avec Dropout Récurrent (RDO)
models['GRU-1-RDO'] = Sequential([
    # Ajout du recurrent_dropout (RDO) pour régulariser les connexions internes du GRU
    GRU(64, recurrent_dropout=0.2, input_shape=(window_size, num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 2. GRU Profond avec RDO et L2
models['GRU-2-RDO-L2'] = Sequential([
    # RDO et régularisation L2 sur les poids pour éviter l'overfitting
    GRU(128, return_sequences=True, recurrent_dropout=0.2, kernel_regularizer=tf.keras.regularizers.l2(1e-4), input_shape=(window_size, num_features)),
    Dropout(0.2),
    GRU(64),
    Dense(32, activation='relu'),
    Dense(1)
])

# 3. GRU-Dense avec Dropout accru et LeakyReLU
models['GRU-Dense-D03'] = Sequential([
    GRU(64, input_shape=(window_size, num_features)),
    Dense(64, activation='relu'),
    Dropout(0.3), # Taux de Dropout accru
    Dense(32, activation=tf.keras.layers.LeakyReLU(alpha=0.01)), # Activation LeakyReLU
    Dense(1)
])

# 4. GRU Simple avec Layer Normalization
models['GRU-1-Norm'] = Sequential([
    # Layer Normalization pour stabiliser les activations
    GRU(64, input_shape=(window_size, num_features)),
    tf.keras.layers.LayerNormalization(),
    Dense(32, activation='relu'),
    Dense(1)
])

# 5. GRU Profond avec sortie Tanh
models['GRU-2-Tanh'] = Sequential([
    GRU(128, return_sequences=True, input_shape=(window_size, num_features)),
    Dropout(0.2),
    GRU(64),
    Dense(32, activation='relu'),
    # Tanh sur la couche Dense de sortie (peut aider à normaliser l'échelle des activations)
    Dense(1, activation='tanh')
])

# --- Améliorations BiDirectionnelles (BiGRU, CNN2-BiGRU, CNN2-BiLSTM) ---

# 6. BiGRU Empilé (Double Bidirectional)
models['BiGRU-Deep'] = Sequential([
    # BiGRU empilé pour une meilleure capture du contexte bidirectionnel
    Bidirectional(GRU(128, return_sequences=True)),
    Dropout(0.2),
    Bidirectional(GRU(64)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 7. CNN2-BiGRU avec RDO
models['CNN2-BiGRU-RDO'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    # RDO dans le BiGRU
    Bidirectional(GRU(64, recurrent_dropout=0.2)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 8. CNN2-BiLSTM avec Taux de Dropout élevé (0.4)
models['CNN2-BiLSTM-D04'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    Bidirectional(LSTM(64)),
    Dropout(0.4), # Taux de Dropout très élevé
    Dense(32, activation='relu'),
    Dense(1)
])

# 9. CNN2-BiGRU avec plus de Neurones (128)
models['CNN2-BiGRU-128'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    # Augmentation de la capacité du BiGRU à 128
    Bidirectional(GRU(128)),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)
])

# 10. BiGRU avec L2 et RDO
models['BiGRU-L2-RDO'] = Sequential([
    # Double régularisation sur le BiGRU
    Bidirectional(GRU(64, recurrent_dropout=0.2, kernel_regularizer=tf.keras.regularizers.l2(1e-4)), input_shape=(window_size, num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# --- Modèles Hybrides GRU/LSTM et Améliorations CNN ---

# 11. CNN2-BiLSTM avec Dropout sur les couches CNN
models['CNN2-BiLSTM-D-CNN'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Dropout(0.1), # Dropout après la première Conv
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    Bidirectional(LSTM(64)),
    Dropout(0.2),
    Dense(1)
])

# 12. GRU-LSTM Profond
models['GRU-LSTM-Deep'] = Sequential([
    GRU(128, return_sequences=True, input_shape=(window_size, num_features)),
    Dropout(0.2),
    LSTM(64), # Passage de GRU à LSTM pour la deuxième couche
    Dense(32, activation='relu'),
    Dense(1)
])

# 13. CNN-BiGRU avec Noyau Large et RDO
models['CNN-BiGRU-K5'] = Sequential([
    # Noyau plus large (5)
    Conv1D(64, 5, padding='same', activation='relu', input_shape=(window_size, num_features)),
    MaxPooling1D(2),
    # RDO sur le BiGRU
    Bidirectional(GRU(64, recurrent_dropout=0.2)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 14. CNN2-BiLSTM avec Global Max Pooling
models['CNN2-BiLSTM-MaxPool'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    Bidirectional(LSTM(64)), # Attention: Le pooling change la dimension, BiLSTM n'est plus compatible directement.
    # Pour le pooling global, on retire le Bidirectional(LSTM) et on passe directement à Dense:
    Dense(32, activation='relu'),
    Dense(1)
])

# 15. CNN2-BiLSTM avec LeakyReLU
models['CNN2-BiLSTM-Leaky'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    Bidirectional(LSTM(64)),
    Dropout(0.2),
    Dense(32, activation=tf.keras.layers.LeakyReLU(alpha=0.01)),
    Dense(1)
])

# --- Modèles d'Ensemble et Profondeur DENSE ---

# 16. GRU-Dense Extra Profond
models['GRU-Dense-Deep'] = Sequential([
    GRU(64, input_shape=(window_size, num_features)),
    Dense(128, activation='relu'), # Plus large
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)
])

# 17. BiGRU + Attention (Approximation)
# Une couche dense qui simule l'agrégation pondérée des sorties de séquence avant le BiGRU
models['BiGRU-PreAttention'] = Sequential([
    Conv1D(64, 1, activation='relu', input_shape=(window_size, num_features)), # "Pointwise" Conv pour une agrégation
    Bidirectional(GRU(64)),
    Dense(32, activation='relu'),
    Dense(1)
])

# 18. CNN2-BiGRU avec 32 Neurones (Moins de paramètres)
models['CNN2-BiGRU-32N'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    # Réduction à 32 neurones pour un modèle plus léger
    Bidirectional(GRU(32)),
    Dense(16, activation='relu'),
    Dense(1)
])

# 19. GRU-Dense avec RDO sur le GRU
models['GRU-Dense-RDO'] = Sequential([
    GRU(64, recurrent_dropout=0.2, input_shape=(window_size, num_features)),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(1)
])

# 20. CNN2-BiLSTM avec Dropout entre les Couches CNN
models['CNN2-BiLSTM-D-Intra'] = Sequential([
    Conv1D(32, 3, padding='same', activation='relu', input_shape=(window_size, num_features)),
    Dropout(0.1), # Dropout entre les couches Conv
    Conv1D(64, 3, padding='same', activation='relu'),
    MaxPooling1D(2),
    Bidirectional(LSTM(64)),
    Dropout(0.2),
    Dense(1)
])


# ===============================
# 6. Entraînement et benchmark + RMSE
# ===============================
results = {}

for name, model in models.items():
    print(f"=== Entraînement du modèle {name} ===")
    
    model.compile(optimizer=Adam(0.001), loss='mse', metrics=['mae'])
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-5)
    
    history = model.fit(
        X_train, y_train,
        epochs=100,
        batch_size=32,
        validation_data=(X_test, y_test),
        callbacks=[early_stop, reduce_lr],
        verbose=0
    )
    
    y_pred_scaled = model.predict(X_test)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_test_real = scaler_y.inverse_transform(y_test)
    
    y_pred_aligned = np.full_like(y_test_real, np.nan)
    if forecast_horizon <= len(y_pred):
        y_pred_aligned[forecast_horizon-1:] = y_pred[:len(y_pred)-forecast_horizon+1]
    
    rmse = np.sqrt(mean_squared_error(y_test_real[forecast_horizon-1:], y_pred_aligned[forecast_horizon-1:]))
    results[name] = rmse
    
    # Plot prédictions (optionnel si beaucoup de modèles)
    plt.figure(figsize=(12,4))
    plt.plot(y_test_real, label='Prix réel')
    plt.plot(y_pred_aligned, label=f'Prix prédit ({name})')
    plt.title(f'{ticker} - {name} - RMSE={rmse:.2f}')
    plt.xlabel('Index')
    plt.ylabel('Prix')
    plt.legend()
    plt.grid(True)
    plt.show()

# ===============================
# 7. Graphique comparatif RMSE
# ===============================
plt.figure(figsize=(12,5))
names = list(results.keys())
values = [results[n] for n in names]
plt.barh(names, values, color='skyblue')
plt.xlabel("RMSE")
plt.title(f"Comparaison des RMSE pour {ticker}")
plt.gca().invert_yaxis()  # le meilleur en haut
plt.grid(axis='x')
plt.show()

# ===============================
# 8. Meilleur modèle + prochaine prédiction
# ===============================
best_model_name = min(results, key=results.get)
print(f"Meilleur modèle : {best_model_name} avec RMSE = {results[best_model_name]:.4f}")

best_model = models[best_model_name]
last_window = X_scaled[-1].reshape(1, window_size, num_features)
pred_scaled = best_model.predict(last_window)
pred_close = scaler_y.inverse_transform(pred_scaled.reshape(-1,1))[0][0]

print(f"Prix prédit pour {ticker} dans {forecast_horizon} jours : {pred_close:.2f}")
```

```python
# ===============================
# 1. Récupération du ticker
# ===============================
ticker = input("Entrez le ticker de l'action ou crypto (ex: AAPL, TSLA, BTC-USD): ")

end_date = datetime.date.today()
start_date = end_date - datetime.timedelta(days=10*365)

data = yf.download(ticker, start=start_date, end=end_date, progress=False).dropna()
data.sort_index(inplace=True)
df = data
print(df.head())

# === 1b. Ajout des indicateurs techniques ===
# SMA sur 10 jours
df['SMA_10'] = df['Close'].rolling(10).mean()
# EMA sur 10 jours
df['EMA_10'] = df['Close'].ewm(span=10, adjust=False).mean()
# RSI sur 14 jours
delta = df['Close'].diff()
up = delta.clip(lower=0)
down = -1*delta.clip(upper=0)
roll_up = up.rolling(14).mean()
roll_down = down.rolling(14).mean()
RS = roll_up / roll_down
df['RSI_14'] = 100 - (100 / (1 + RS))

df.dropna(inplace=True)  # supprimer les lignes avec NaN dues aux indicateurs

# === 2. Création des séquences multivariées ===
def create_windows(df, window_size, forecast_horizon, feature_cols):
    X, y = [], []
    for i in range(len(df) - window_size - forecast_horizon + 1):
        window = df.iloc[i:i+window_size][feature_cols].values
        target = df.iloc[i + window_size + forecast_horizon - 1]["Close"]
        X.append(window)
        y.append(target)
    return np.array(X), np.array(y)

feature_cols = ["Open","High","Low","Close","Volume","SMA_10","EMA_10","RSI_14"]
num_features = len(feature_cols)
window_size = 20       # nombre de jours d’historique
forecast_horizon = 1   # prédire le Close dans X jours

X, y = create_windows(df, window_size, forecast_horizon, feature_cols)

# ===============================
# 3. Normalisation
# ===============================
X_scaled = np.zeros_like(X)
scalers_X = {}
for i in range(num_features):
    scalers_X[i] = MinMaxScaler()
    X_scaled[:,:,i] = scalers_X[i].fit_transform(X[:,:,i])

scaler_y = MinMaxScaler()
y_scaled = scaler_y.fit_transform(y.reshape(-1,1))

# ===============================
# 4. Split train/test
# ===============================
split = int(0.95 * len(X_scaled))
X_train, X_test = X_scaled[:split], X_scaled[split:]
y_train, y_test = y_scaled[:split], y_scaled[split:]


# ===============================================================================
# 🚀 VARIANTES OPTIMISÉES DE TCN-BIGRU
# ===============================================================================

models = {}

# C. Modèle d'Ensemble (Architecture Parallèle)
# On crée un modèle qui prend les deux forces : Convolutions (TCN) et Séquentiel (GRU)
def create_ensemble():
    inputs = Input(shape=(window_size, num_features))
    
    # Branche TCN (60% influence théorique)
    tcn_out = TCN(nb_filters=32, kernel_size=2, dilations=[1, 2], return_sequences=False)(inputs)
    
    # Branche GRU (40% influence théorique)
    gru_out = Bidirectional(GRU(32))(inputs)
    
    concat = tf.keras.layers.Concatenate()([tcn_out, gru_out])
    dense = Dense(16, activation='relu')(concat)
    output = Dense(1)(dense)
    
    return Model(inputs=inputs, outputs=output)

models['Ensemble-TCN-BiGRU'] = create_ensemble()

def create_attention_tcn_gru():
    inputs = Input(shape=(window_size, num_features))
    
    # Branche 1 : TCN
    # Correction : On utilise 'dropout_rate' au lieu de 'dropout'
    tcn_out = TCN(nb_filters=64, 
                  kernel_size=3, 
                  dilations=[1, 2, 4], 
                  return_sequences=True, 
                  dropout_rate=0.1)(inputs)
    
    # Branche 2 : BiGRU
    gru_out = Bidirectional(GRU(64, return_sequences=True))(inputs)
    
    # Mécanisme d'Attention
    # L'attention compare les deux branches pour extraire les caractéristiques saillantes
    query = Dense(128)(tcn_out)
    value = Dense(128)(gru_out)
    attn_out = Attention()([query, value])
    
    # Global Pooling pour réduire la dimension temporelle avant la sortie
    # Cela permet de transformer la séquence (window_size, 128) en un vecteur fixe (128)
    avg_pool = GlobalAveragePooling1D()(attn_out)
    
    dense = Dense(32, activation='leaky_relu')(avg_pool)
    dense = Dropout(0.2)(dense)
    output = Dense(1)(dense)
    
    model = Model(inputs=inputs, outputs=output)
    
    return model

models['Ensemble-Attention-TCN-BiGRU'] = create_attention_tcn_gru()

models['CNN2-BiGRU'] = Sequential([
    Conv1D(32,3,padding='same',activation='relu',input_shape=(window_size,num_features)),
    Conv1D(64,3,padding='same',activation='relu'),
    MaxPooling1D(2),
    Bidirectional(GRU(64)),
    Dense(32, activation='relu'),
    Dense(1)
])

models['BiGRU-L2-RDO'] = Sequential([
    # Double régularisation sur le BiGRU
    Bidirectional(GRU(64, recurrent_dropout=0.2, kernel_regularizer=tf.keras.regularizers.l2(1e-4)), input_shape=(window_size, num_features)),
    Dense(32, activation='relu'),
    Dense(1)
])

# ===============================
# 6. Entraînement et benchmark + RMSE
# ===============================
results = {}

for name, model in models.items():
    print(f"=== Entraînement du modèle {name} ===")
    
    model.compile(optimizer=Adam(0.001), loss=tf.keras.losses.Huber(), metrics=['mae'])
    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-5)
    
    history = model.fit(
        X_train, y_train,
        epochs=100,
        batch_size=32,
        validation_data=(X_test, y_test),
        callbacks=[early_stop, reduce_lr],
        verbose=0
    )
    
    y_pred_scaled = model.predict(X_test)
    y_pred = scaler_y.inverse_transform(y_pred_scaled)
    y_test_real = scaler_y.inverse_transform(y_test)
    
    y_pred_aligned = np.full_like(y_test_real, np.nan)
    if forecast_horizon <= len(y_pred):
        y_pred_aligned[forecast_horizon-1:] = y_pred[:len(y_pred)-forecast_horizon+1]
    
    rmse = np.sqrt(mean_squared_error(y_test_real[forecast_horizon-1:], y_pred_aligned[forecast_horizon-1:]))
    results[name] = rmse
    
    # Plot prédictions (optionnel si beaucoup de modèles)
    plt.figure(figsize=(12,4))
    plt.plot(y_test_real, label='Prix réel')
    plt.plot(y_pred_aligned, label=f'Prix prédit ({name})')
    plt.title(f'{ticker} - {name} - RMSE={rmse:.2f}')
    plt.xlabel('Index')
    plt.ylabel('Prix')
    plt.legend()
    plt.grid(True)
    plt.show()

# ===============================
# 7. Graphique comparatif RMSE
# ===============================
plt.figure(figsize=(12,5))
names = list(results.keys())
values = [results[n] for n in names]
plt.barh(names, values, color='skyblue')
plt.xlabel("RMSE")
plt.title(f"Comparaison des RMSE pour {ticker}")
plt.gca().invert_yaxis()  # le meilleur en haut
plt.grid(axis='x')
plt.show()

# ===============================
# 8. Meilleur modèle + prochaine prédiction
# ===============================
best_model_name = min(results, key=results.get)
print(f"Meilleur modèle : {best_model_name} avec RMSE = {results[best_model_name]:.4f}")

best_model = models[best_model_name]
last_window = X_scaled[-1].reshape(1, window_size, num_features)
pred_scaled = best_model.predict(last_window)
pred_close = scaler_y.inverse_transform(pred_scaled.reshape(-1,1))[0][0]

print(f"Prix prédit pour {ticker} dans {forecast_horizon} jours : {pred_close:.2f}")
```
